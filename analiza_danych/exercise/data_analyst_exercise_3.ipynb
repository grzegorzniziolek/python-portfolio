```python
display(city_df.head(3))
display(participants_df.head(3))
display(trials_df.head(3))
```

```python
display(city_df.dtypes)
display(participants_df.dtypes)
display(trials_df.dtypes)
```

```python
dfs = {
    "participants_df": participants_df,
    "trials_df": trials_df,
    "city_df": city_df
}

typy_kolumn = {
    
    "participants_df": 
    {
        "numeric": ["sleep_hours", "stress_level", "reaction_time_ms", "memory_score", "attention_score"],
        "date": ["session_date"],
        "comma": ["reaction_time_ms", "memory_score", "attention_score"]
    },


    "trials_df": 
    {
        "numeric": ["rt_ms", "confidence_1_5"],
        "date": [],
        "comma": ["rt_ms"]
    },


    "city_df": 
    {
        "numeric": [],
        "date": [],
        "comma": []
    }
}

for nazwa_df, df in dfs.items():
    
    for col in typy_kolumn[nazwa_df]["comma"]:
        df[col] = df[col].astype(str).str.replace(",", ".")

    for col in typy_kolumn[nazwa_df]["numeric"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")

    for col in typy_kolumn[nazwa_df]["date"]:
        df[col] = pd.to_datetime(df[col], errors="coerce")
```
